```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.DataFrame({
    'EmpID': [1, 2, 3, 4, 5],
    'Name': ['Nandha', 'George', 'Daniel', 'Wesley', 'Rita'],
    'Dept': ['HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Salary': [40000, 50000, 60000, 42000, 48000],
    'Experience': [3, 5, 4, 2, 4],
    'Join_Date': ['2020-01-01','2021-05-12','2021-08-20','2022-02-10','2021-09-01']
})
```

```python
from scipy import stats
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Join_Date']=pd.to_datetime(df['Join_Date'])
df['Bonus']=df['Salary']*0.1
df['Total_Comp']=df['Salary']+df['Bonus']
df['High_Salary']=np.where(df['Salary']>50000,'Yes', 'No')
df['Dept']=le.fit_transform(df['Dept'])
df['Join_Year']=df['Join_Date'].dt.year
df['Join_Month']=df['Join_Date'].dt.month
df['Salary_scaled']=(df['Salary']-df['Salary'].min())/(df['Salary'].max()-df['Salary'].min())
df['Experience_z']=np.abs(stats.zscore(df['Experience']))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y1 = x * 2
y2 = x ** 2
y3 = np.sqrt(x)
y4 = x ** 3

plt.figure(figsize=(10,6))

plt.subplot(2,2,1)
plt.plot(x, y1, color='red')
plt.title('Linear')

plt.subplot(2,2,2)
plt.plot(x, y2, color='green')
plt.title('Square')

plt.subplot(2,2,3)
plt.plot(x, y3, color='orange')
plt.title('Square Root')

plt.subplot(2,2,4)
plt.plot(x, y4, color='blue')
plt.title('Cubic')

plt.tight_layout()
plt.show()
```
